```python
# ============================================================
# 01. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np

from sqlalchemy import create_engine

print("Libraries imported successfully!")
```

```python
# ============================================================
# 02. LOAD RAW DATA
# ============================================================

file_path = r"D:\data_analyst_road_map\project\customer_churn\dataset\raw\customer_churn_dataset-training-master.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
# ============================================================
# 03. DATASET INFORMATION
# ============================================================

df.info()
```

```python
# ============================================================
# 04. CHECK MISSING VALUES
# ============================================================

missing_values = df.isnull().sum()

print("Missing values by column:")
print(missing_values)

print("\nTotal missing values:", missing_values.sum())
```

```python
# ============================================================
# 05. CHECK COMPLETELY EMPTY ROWS
# ============================================================

empty_rows = df[df.isnull().all(axis=1)]

print("Completely empty rows:", len(empty_rows))

if len(empty_rows) > 0:
    print("\nEmpty rows:")
    print(empty_rows)
```

```python
# ============================================================
# 06. REMOVE COMPLETELY EMPTY ROWS
# ============================================================

rows_before = len(df)

df = df.dropna(how="all").copy()

rows_after = len(df)

print("Rows before:", rows_before)
print("Rows after:", rows_after)
print("Rows removed:", rows_before - rows_after)
```

```python
# ============================================================
# 07. CHECK DUPLICATE ROWS
# ============================================================

duplicate_rows = df.duplicated().sum()

print("Duplicate rows:", duplicate_rows)
```

```python
# ============================================================
# 08. CHECK DUPLICATE CUSTOMER ID
# ============================================================

duplicate_customer_id = df["CustomerID"].duplicated().sum()

print("Duplicate CustomerID:", duplicate_customer_id)
```

```python
# ============================================================
# 09. CHECK CATEGORICAL VALUES
# ============================================================

categorical_columns = [
    "Gender",
    "Subscription Type",
    "Contract Length"
]

for col in categorical_columns:
    print(f"\n{col}:")
    print(df[col].unique())
```

```python
# ============================================================
# 10. CLEAN CATEGORICAL VALUES
# ============================================================

for col in categorical_columns:
    df[col] = (
        df[col]
        .astype("string")
        .str.strip()
    )

print("Categorical values cleaned successfully!")
```

```python
# ============================================================
# 11. CONVERT NUMERIC COLUMNS
# ============================================================

numeric_columns = [
    "Age",
    "Tenure",
    "Usage Frequency",
    "Support Calls",
    "Payment Delay",
    "Total Spend",
    "Last Interaction",
    "Churn"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

print("Numeric columns converted successfully!")
```

```python
# ============================================================
# 12. CONVERT INTEGER COLUMNS
# ============================================================

integer_columns = [
    "Age",
    "Tenure",
    "Usage Frequency",
    "Support Calls",
    "Payment Delay",
    "Last Interaction",
    "Churn"
]

for col in integer_columns:
    df[col] = df[col].astype("Int64")

print("Integer columns converted successfully!")
```

```python
# ============================================================
# 13. FORMAT TOTAL SPEND
# ============================================================

df["Total Spend"] = df["Total Spend"].astype("float64")

df["Total Spend"] = df["Total Spend"].round(2)

print(df["Total Spend"].head())
print("\nData type:", df["Total Spend"].dtype)
```

```python
# ============================================================
# 14. CHECK NUMERIC RANGES
# ============================================================

print(df[numeric_columns].describe().T)
```

```python
# ============================================================
# 15. CHECK INVALID VALUES
# ============================================================

valid_ranges = {
    "Age": (18, 65),
    "Tenure": (1, 60),
    "Usage Frequency": (1, 30),
    "Support Calls": (0, 10),
    "Payment Delay": (0, 30),
    "Last Interaction": (1, 30),
    "Churn": (0, 1)
}

for col, (min_value, max_value) in valid_ranges.items():

    invalid = df[
        (df[col] < min_value) |
        (df[col] > max_value)
    ]

    print(
        f"{col}: {len(invalid)} invalid values "
        f"(expected {min_value}-{max_value})"
    )
```

```python
# ============================================================
# 16. CHECK NEGATIVE TOTAL SPEND
# ============================================================

negative_spend = df[df["Total Spend"] < 0]

print("Negative Total Spend:", len(negative_spend))

if len(negative_spend) > 0:
    print(negative_spend)
```

```python
# ============================================================
# 17. CHECK MISSING VALUES AFTER TYPE CONVERSION
# ============================================================

missing_after_conversion = df.isnull().sum()

print(missing_after_conversion)

print(
    "\nTotal missing values:",
    missing_after_conversion.sum()
)
```

```python
# ============================================================
# 18. REMOVE REMAINING MISSING VALUES
# ============================================================

rows_before = len(df)

df = df.dropna().copy()

rows_after = len(df)

print("Rows before:", rows_before)
print("Rows after:", rows_after)
print("Rows removed:", rows_before - rows_after)
```

```python
# ============================================================
# 19. CONVERT COLUMN NAMES TO SNAKE_CASE
# ============================================================

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)

print("Column names:")
print(df.columns.tolist())
```

```python
# ============================================================
# 20. FINAL DATA QUALITY CHECK
# ============================================================

print("=" * 60)
print("FINAL DATA QUALITY CHECK")
print("=" * 60)

print("\nShape:")
print(df.shape)

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nDuplicate Customer ID:")
print(df["customerid"].duplicated().sum())

print("\nData types:")
print(df.dtypes)
```

```python
# ============================================================
# 21. CHECK CHURN DISTRIBUTION
# ============================================================

churn_counts = df["churn"].value_counts()

print("Churn counts:")
print(churn_counts)

print("\nChurn percentage:")
print(
    (df["churn"].value_counts(normalize=True) * 100)
    .round(2)
)
```

```python
# ============================================================
# 22. FINAL CATEGORICAL CHECK
# ============================================================

categorical_columns_cleaned = [
    "gender",
    "subscription_type",
    "contract_length"
]

for col in categorical_columns_cleaned:
    print(f"\n{col}:")
    print(df[col].value_counts())
```

```python
# ============================================================
# 23. FINAL NUMERIC CHECK
# ============================================================

numeric_columns_cleaned = [
    "age",
    "tenure",
    "usage_frequency",
    "support_calls",
    "payment_delay",
    "total_spend",
    "last_interaction",
    "churn"
]

print(df[numeric_columns_cleaned].describe().T)
```

```python
# ============================================================
# 24. EXPORT CLEANED DATASET
# ============================================================

output_path = r"D:\data_analyst_road_map\project\customer_churn\dataset\cleaned\customer_churn_cleaned.csv"

df.to_csv(
    output_path,
    index=False
)

print("Cleaned dataset exported successfully!")
print("File:", output_path)
print("Rows:", len(df))
print("Columns:", len(df.columns))
```

```python
# ============================================================
# 25. CONNECT TO POSTGRESQL
# ============================================================

from sqlalchemy import create_engine

username = "postgres"
password = "12345"
host = "localhost"
port = "5432"
database = "customer_churn"

connection_string = (
    f"postgresql+psycopg2://"
    f"{username}:{password}@"
    f"{host}:{port}/{database}"
)

engine = create_engine(connection_string)

print("PostgreSQL connection created successfully!")
```

```python
# ============================================================
# 26. TEST POSTGRESQL CONNECTION
# ============================================================

from sqlalchemy import text

with engine.connect() as connection:
    result = connection.execute(
        text("SELECT version();")
    )

    version = result.fetchone()[0]

print("PostgreSQL connection successful!")
print("\nPostgreSQL version:")
print(version)
```

```python
# ============================================================
# 27. LOAD DATA INTO POSTGRESQL
# ============================================================

from sqlalchemy.types import String, Integer, Numeric

table_name = "customer_churn"

df.to_sql(
    table_name,
    engine,
    if_exists="replace",
    index=False,
    dtype={
        "customerid": String(50),
        "age": Integer(),
        "gender": String(20),
        "tenure": Integer(),
        "usage_frequency": Integer(),
        "support_calls": Integer(),
        "payment_delay": Integer(),
        "subscription_type": String(20),
        "contract_length": String(20),
        "total_spend": Numeric(12, 2),
        "last_interaction": Integer(),
        "churn": Integer()
    }
)

print("Data loaded successfully!")
print("Table:", table_name)
print("Rows:", len(df))
```

```python
# ============================================================
# 28. VERIFY POSTGRESQL TABLE
# ============================================================

query = """
SELECT *
FROM customer_churn
LIMIT 10;
"""

df_postgres = pd.read_sql(
    query,
    engine
)

df_postgres
```

```python
# ============================================================
# 29. CHECK ROW COUNT
# ============================================================

query = """
SELECT COUNT(*) AS total_customers
FROM customer_churn;
"""

result = pd.read_sql(
    query,
    engine
)

result
```

```python
# ============================================================
# 30. CHECK POSTGRESQL DATA TYPES
# ============================================================

query = """
SELECT
    column_name,
    data_type
FROM information_schema.columns
WHERE table_name = 'customer_churn'
ORDER BY ordinal_position;
"""

postgres_schema = pd.read_sql(
    query,
    engine
)

postgres_schema
```

```python
# ============================================================
# 31. FINAL POSTGRESQL VALIDATION
# ============================================================

query = """
SELECT
    COUNT(*) AS total_rows,
    COUNT(customerid) AS customerid_count,
    COUNT(*) - COUNT(customerid) AS missing_customerid,
    COUNT(*) - COUNT(churn) AS missing_churn
FROM customer_churn;
"""

final_check = pd.read_sql(
    query,
    engine
)

final_check
```
